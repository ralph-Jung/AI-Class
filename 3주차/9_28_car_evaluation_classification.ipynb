```python
# 1. Logistic Regression - Car Evaluation 데이터
import pandas as pd

from sklearn.model_selection import train_test_split
# 의사결정나무 분류 모델. 데이터를 조건에 따라 가지치기하듯 나눠서 분류하는 알고리즘.
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report
from sklearn.preprocessing import LabelEncoder



# 헤더가 없는 파일이므로 header=None 으로 읽고 열 이름을 직접 지정
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/car.data.csv",
                 header=None, names=columns)

feature_cols = ["buying", "maint", "doors", "persons", "lug_boot", "safety"]

label_encoders = {}
for column in feature_cols:
    label_encoders[column] = LabelEncoder()
    df[column] = label_encoders[column].fit_transform(df[column].astype(str))


# X: class 를 제외한 나머지 열 (입력 데이터)
X = df.drop("class", axis=1)
# y: 차량 평가 등급 (정답 데이터: unacc / acc / good / vgood)
y = df["class"]

# 80% 학습, 20% 테스트
# stratify=y : good, vgood 처럼 개수가 적은 클래스도 비율대로 나눠지게 함
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

# model 이라는 변수에 내가 원하는 모델을 추가한다
# 의사결정트리
model = DecisionTreeClassifier()

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print()
print(classification_report(y_test, y_pred))
```
