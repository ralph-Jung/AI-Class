```python
# 1. Decision Tree
import pandas as pd

from sklearn.model_selection import train_test_split
# 의사결정나무 분류 모델. 데이터를 조건에 따라 가지치기하듯 나눠서 분류하는 알고리즘.
from sklearn.tree import DecisionTreeClassifier
# 예측값이 실제값과 얼마나 맞았는지 정확도(맞춘 비율)를 계산 / 클래스별로 precision(정밀도), recall(재현율), f1-score 등을 표로 정리해서 보여줌
from sklearn.metrics import accuracy_score, classification_report


# 데이터에는 데이터 프레임(액셀 형식이 데이터프레임 이다) 과 넘파이 로 나눈다
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/iris.csv")


# x에는 species 를 제외한 나머지 열들을 가져옴 ( 입력 데이터 )
X = df.drop("species", axis=1)
# y에는 품종  ( 정답 데이터 )
y = df["species"]


# 전체 데이터를 80% 는 학습용으로 사용하고 나머지 20% 는 테스트 용으로 나누기
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# model 이라는 변수에 내가 원하는 모델을 추가한다
# 의사결정트리
model = DecisionTreeClassifier()

# model 을 학습 데이터를 바탕으로 학습 시킨다
model.fit(X_train, y_train)

# 내가 위에서 테스트 데이터 20% 만들어 놓은 것을 가지고 얼마나 성능이 나오는지 예측
# 여기 y_pred 에는 X_test를 모델에 넣었을 때 나온 y 값들이 나오고 이 예측값을 정답인 y_pred 랑 비교를 해서 몇개를 맞췄는지 정확도를 매긴다
y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))

print()

print(classification_report(y_test, y_pred))



```

```python
# 3. Logistic Regression
import pandas as pd

from sklearn.model_selection import train_test_split
# 의사결정나무 분류 모델. 데이터를 조건에 따라 가지치기하듯 나눠서 분류하는 알고리즘.
from sklearn.linear_model import LogisticRegression
# 예측값이 실제값과 얼마나 맞았는지 정확도(맞춘 비율)를 계산 / 클래스별로 precision(정밀도), recall(재현율), f1-score 등을 표로 정리해서 보여줌
from sklearn.metrics import accuracy_score, classification_report


# 데이터에는 데이터 프레임(액셀 형식이 데이터프레임 이다) 과 넘파이 로 나눈다
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/iris.csv")


# x에는 species 를 제외한 나머지 열들을 가져옴 ( 입력 데이터 )
X = df.drop("species", axis=1)
# y에는 품종  ( 정답 데이터 )
y = df["species"]


# 전체 데이터를 80% 는 학습용으로 사용하고 나머지 20% 는 테스트 용으로 나누기
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# model 이라는 변수에 내가 원하는 모델을 추가한다
# 의사결정트리
model = LogisticRegression(max_iter=200)

# model 을 학습 데이터를 바탕으로 학습 시킨다
model.fit(X_train, y_train)

# 내가 위에서 테스트 데이터 20% 만들어 놓은 것을 가지고 얼마나 성능이 나오는지 예측
# 여기 y_pred 에는 X_test를 모델에 넣었을 때 나온 y 값들이 나오고 이 예측값을 정답인 y_pred 랑 비교를 해서 몇개를 맞췄는지 정확도를 매긴다
y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))

print()

print(classification_report(y_test, y_pred))
```
