```python
# 3. Logistic Regression
import pandas as pd

from sklearn.model_selection import train_test_split
# 의사결정나무 분류 모델. 데이터를 조건에 따라 가지치기하듯 나눠서 분류하는 알고리즘.
from sklearn.neighbors import KNeighborsRegressor
# 예측값이 실제값과 얼마나 맞았는지 정확도(맞춘 비율)를 계산 / 클래스별로 precision(정밀도), recall(재현율), f1-score 등을 표로 정리해서 보여줌
from sklearn.metrics import mean_squared_error

df=pd.read_csv("/content/drive/MyDrive/Colab Notebooks/diabetes.csv")

x=df.drop(["BMI", "Outcome"],axis=1)
X=df.drop("BMI",axis=1)


# 여기서 학습데이터 랑 테스트데이터를 8:2 로 나눌 때 무작위로 나누기 때문에 실행할 때 마다 정확도가 달라진다
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# model 이라는 변수에 내가 원하는 모델을 추가한다
# 의사결정트리
model = KNeighborsRegressor()

# model 을 학습 데이터를 바탕으로 학습 시킨다
model.fit(X_train, y_train)

# 내가 위에서 테스트 데이터 20% 만들어 놓은 것을 가지고 얼마나 성능이 나오는지 예측
# 여기 y_pred 에는 X_test를 모델에 넣었을 때 나온 y 값들이 나오고 이 예측값을 정답인 y_pred 랑 비교를 해서 몇개를 맞췄는지 정확도를 매긴다
dy_preds = model.predict(X_test)

print('평균제곱근오차',mean_squared_error(dy_preds,y_test))
```
